## Exercise 1

Tenemos las siguientes temperaturas en Celsius:

[12, 18, 21, 25, 30, 35, 40]

Utilizando NumPy:

1. Crear un ndarray.
2. Convertir las temperaturas a Fahrenheit.
3. Encontrar la temperatura máxima.
4. Encontrar la temperatura mínima.
5. Calcular el promedio.
6. Seleccionar las temperaturas mayores a 25°C.
7. Seleccionar las temperaturas entre 20°C y 35°C.
8. Ordenar las temperaturas.
9. Convertir el array en un shape de 1 × 7.
10. Convertirlo nuevamente en un array de 7 × 1.

```python
import numpy as np

temps = [12, 18, 21, 25, 30, 35, 40]
n_temps = np.array(temps)
print(n_temps)
print(type(n_temps))

n_tempsf = n_temps * (9/5) + 32
print(n_tempsf)

print(n_temps.max())
print(n_temps.min())
print(n_temps.mean())

print(n_temps[n_temps > 25])
print(n_temps[(n_temps > 20) & (n_temps < 35)])

print(np.sort(n_temps))

print(n_temps.reshape(1, 7))

print(n_temps.reshape(7, 1))

print(n_temps.reshape(7))
```

## Excercise 2

Tenemos las ventas de una empresa:

           Enero Febrero Marzo
Producto A   10    20     30
Producto B   15    25     35
Producto C    5    15     25

Crear el array y calcular:

1. Ventas totales.
2. Ventas por producto.
3. Ventas por mes.
4. Producto con mayores ventas.
5. Mes con mayores ventas.
6. Todas las ventas mayores a 20.
7. Promedio de ventas.

```python
import string
import numpy as np

letters = string.ascii_uppercase
months = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]
sales = [[10, 20, 30],
         [15, 25, 35],
         [ 5, 15, 25]]

n_sales = np.array(sales)
print(n_sales)

print("Ventas totales", np.sum(n_sales))
max_per_product = np.sum(n_sales, axis=1)
print("Ventas por producto", max_per_product)
max_per_month = np.sum(n_sales, axis=0)
print("Ventas por mes", np.sum(n_sales, axis=0))

max_product_sales = np.max(max_per_product)
max_product = np.argmax(max_per_product)
max_product_ref = letters[max_product]
print("Producto con mayores ventas", max_product_sales)
print("Producto con mayores ventas", max_product)
print("Producto con mayores ventas", max_product_ref)

max_month_sales = np.max(max_per_month)
max_month = np.argmax(max_per_month)
max_month_ref = months[max_month]
print("Mes con mayores ventas", max_month_sales)
print("Mes con mayores ventas", max_month)
print("Mes con mayores ventas", max_month_ref)

i_gt_20 = np.where(n_sales > 20)
print("Ventas mayores a 20", n_sales[i_gt_20])

print("Promedio de todas las ventas", np.mean(n_sales))
```
