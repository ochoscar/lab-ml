# 2. The Matplotlib Mental Model

The most important concept in Matplotlib is understanding the difference between:

- **Figure** — the entire canvas
- **Axes** — one plotting area inside the figure
- **Axis** — the x-axis or y-axis object
- **Artist** — visual elements such as lines, text, legends, patches, etc.

The common pattern is:

```python
fig, ax = plt.subplots()
ax.plot(x, y)
plt.show()
```

For serious work, prefer the **object-oriented API** (`fig, ax = plt.subplots()`) because
it makes complex figures easier to control.

````python
fig, ax = plt.subplots()

ax.plot([1, 2, 3, 4], [10, 20, 15, 30])

plt.show()
````

# 4. Line Charts

````python
x = np.arange(0, 10)
y = x ** 2

fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_title("Quadratic Function")
ax.set_xlabel("X")
ax.set_ylabel("X²")
plt.show()
````

Line charts are ideal for:

- trends over time
- continuous measurements
- comparing trajectories
- mathematical functions

````python
x = np.linspace(0, 2 * np.pi, 200)
y = np.sin(x)

fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_title("Sine Wave")
ax.set_xlabel("x")
ax.set_ylabel("sin(x)")
plt.show()
````

# 5. Multiple Lines

````python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
product_a = [120, 150, 140, 180, 210, 230]
product_b = [100, 130, 160, 155, 190, 220]

fig, ax = plt.subplots()

ax.plot(months, product_a, label="Product A")
ax.plot(months, product_b, label="Product B")

ax.set_title("Monthly Sales")
ax.set_xlabel("Month")
ax.set_ylabel("Sales")
ax.legend()

plt.show()
````

The `label=` argument creates entries for the legend. `ax.legend()` displays them.

# 6. Markers and Line Styles

````python
x = np.arange(1, 8)

fig, ax = plt.subplots()

ax.plot(x, x * 2, marker="o", linestyle="-", label="Solid")
ax.plot(x, x * 2 + 2, marker="s", linestyle="--", label="Dashed")
ax.plot(x, x * 2 + 4, marker="^", linestyle=":", label="Dotted")

ax.legend()
plt.show()
````

Common options:

### Markers
- `o` circle
- `s` square
- `^` triangle
- `D` diamond
- `x` x-marker

### Line styles
- `-` solid
- `--` dashed
- `:` dotted
- `-.` dash-dot

# 7. Titles, Labels, and Legends

````python
x = np.arange(1, 6)
sales = [100, 140, 120, 180, 210]

fig, ax = plt.subplots()

ax.plot(x, sales, marker="o", label="Sales")

ax.set_title("Sales Performance", fontsize=16)
ax.set_xlabel("Month", fontsize=12)
ax.set_ylabel("Revenue ($)", fontsize=12)
ax.legend(title="Metric")

plt.show()
````

# 8. Limits, Ticks, and Scales

````python
x = np.arange(1, 11)
y = x ** 2

fig, ax = plt.subplots()

ax.plot(x, y)
ax.set_xlim(0, 11)
ax.set_ylim(0, 110)

ax.set_xticks([1, 3, 5, 7, 9])
ax.set_yticks([0, 25, 50, 75, 100])

plt.show()
````

Matplotlib also supports logarithmic scales:

```python
ax.set_xscale("log")
ax.set_yscale("log")
```

Use logarithmic scales only when they help the reader understand the data.

````python
x = np.logspace(0, 4, 100)
y = x ** 2

fig, ax = plt.subplots()
ax.plot(x, y)
ax.set_xscale("log")
ax.set_yscale("log")
ax.set_title("Log-Log Plot")
plt.show()
````

# 9. Grids and Reference Lines

````python
x = np.arange(1, 11)
y = x ** 2

fig, ax = plt.subplots()
ax.plot(x, y)

ax.grid(True, alpha=0.3)
ax.axhline(50, linestyle="--")
ax.axvline(5, linestyle=":")
ax.axhspan(40, 60, alpha=0.15)

plt.show()
````

Useful reference functions:

- `ax.axhline()` — horizontal reference line
- `ax.axvline()` — vertical reference line
- `ax.axhspan()` — horizontal shaded region
- `ax.axvspan()` — vertical shaded region

# 10. Colors and Transparency

````python
x = np.arange(1, 11)

fig, ax = plt.subplots()

ax.plot(x, x, color="tab:blue", linewidth=2)
ax.plot(x, x ** 1.5, color="tab:orange", alpha=0.6, linewidth=3)

plt.show()
````

Important visual properties:

- `color`
- `alpha`
- `linewidth`
- `linestyle`
- `marker`
- `markersize`
- `markerfacecolor`
- `markeredgecolor`

Use color to communicate meaning, not merely decoration.

# 11. Bar Charts

````python
products = ["Laptop", "Monitor", "Keyboard", "Mouse"]
sales = [120, 180, 90, 220]

fig, ax = plt.subplots()

ax.bar(products, sales)
ax.set_title("Sales by Product")
ax.set_ylabel("Units Sold")

plt.show()
````

Bar charts are excellent for comparing discrete categories.

Keep category labels readable and avoid unnecessary 3D effects.

# 12. Horizontal Bar Charts

````python
products = ["Laptop", "Monitor", "Keyboard", "Mouse"]
sales = [120, 180, 90, 220]

fig, ax = plt.subplots()

ax.barh(products, sales)
ax.set_title("Sales by Product")
ax.set_xlabel("Units Sold")

plt.show()
````

Horizontal bars are especially useful when category names are long.

# 13. Grouped Bar Charts

````python
products = ["A", "B", "C", "D"]
online = [100, 130, 90, 160]
store = [80, 120, 110, 140]

x = np.arange(len(products))
width = 0.35

fig, ax = plt.subplots()

ax.bar(x - width / 2, online, width, label="Online")
ax.bar(x + width / 2, store, width, label="Store")

ax.set_xticks(x)
ax.set_xticklabels(products)
ax.set_ylabel("Sales")
ax.set_title("Online vs Store Sales")
ax.legend()

plt.show()
````

# 14. Stacked Bar Charts

````python
products = ["A", "B", "C", "D"]
online = np.array([100, 130, 90, 160])
store = np.array([80, 120, 110, 140])

fig, ax = plt.subplots()

ax.bar(products, online, label="Online")
ax.bar(products, store, bottom=online, label="Store")

ax.set_title("Stacked Sales")
ax.set_ylabel("Sales")
ax.legend()

plt.show()
````

Stacked bars show composition. They become difficult to compare when there are many segments.

# 15. Histograms

````python
np.random.seed(42)
ages = np.random.normal(loc=35, scale=8, size=1000)

fig, ax = plt.subplots()

ax.hist(ages, bins=20, edgecolor="black")

ax.set_title("Age Distribution")
ax.set_xlabel("Age")
ax.set_ylabel("Frequency")

plt.show()
````

Histograms show the distribution of a numerical variable.

Important parameters include:

- `bins`
- `range`
- `density`
- `alpha`
- `edgecolor`

Try changing `bins` and observe how the interpretation changes.

````python
fig, ax = plt.subplots()

ax.hist(ages, bins=10, alpha=0.6, label="10 bins")
ax.hist(ages, bins=30, alpha=0.4, label="30 bins")

ax.legend()
plt.show()
````

# 16. Box Plots

````python
np.random.seed(10)

group_a = np.random.normal(50, 8, 100)
group_b = np.random.normal(60, 12, 100)
group_c = np.random.normal(55, 5, 100)

fig, ax = plt.subplots()

ax.boxplot([group_a, group_b, group_c])
ax.set_xticklabels(["A", "B", "C"])
ax.set_title("Distribution Comparison")
ax.set_ylabel("Value")

plt.show()
````

A box plot summarizes:

- median
- lower and upper quartiles
- spread
- potential outliers

It is useful for comparing distributions across groups.

# 17. Scatter Plots

````python
np.random.seed(42)

x = np.random.normal(50, 10, 100)
y = 2 * x + np.random.normal(0, 15, 100)

fig, ax = plt.subplots()

ax.scatter(x, y, alpha=0.6)

ax.set_title("X vs Y")
ax.set_xlabel("X")
ax.set_ylabel("Y")

plt.show()
````

Scatter plots are useful for investigating relationships between two numerical variables.

Look for:

- direction
- strength
- clusters
- outliers
- non-linear patterns

# 18. Bubble Charts

````python
np.random.seed(5)

x = np.random.uniform(0, 100, 30)
y = np.random.uniform(0, 100, 30)
size = np.random.uniform(50, 500, 30)

fig, ax = plt.subplots()

ax.scatter(x, y, s=size, alpha=0.5)

ax.set_title("Bubble Chart")
ax.set_xlabel("X")
ax.set_ylabel("Y")

plt.show()
````

The `s=` argument controls marker area. Bubble size can encode a third numerical variable.

# 19. Pie Charts

````python
labels = ["Engineering", "Sales", "Marketing", "Support"]
values = [40, 25, 20, 15]

fig, ax = plt.subplots()

ax.pie(
    values,
    labels=labels,
    autopct="%1.1f%%",
    startangle=90
)

ax.set_title("Employees by Department")

plt.show()
````

Pie charts can work for a small number of categories representing parts of a whole.

For many categories or precise comparisons, a bar chart is usually easier to read.

# 20. Subplots

````python
x = np.arange(1, 11)

fig, axes = plt.subplots(2, 2, figsize=(10, 7))

axes[0, 0].plot(x, x)
axes[0, 0].set_title("Linear")

axes[0, 1].plot(x, x ** 2)
axes[0, 1].set_title("Quadratic")

axes[1, 0].plot(x, np.sqrt(x))
axes[1, 0].set_title("Square Root")

axes[1, 1].plot(x, np.log(x))
axes[1, 1].set_title("Logarithm")

plt.show()
````

`plt.subplots()` returns:

```python
fig, axes
```

When there are multiple axes, `axes` is usually a NumPy array of Axes objects.

# 21. Figure Size and Layout

````python
fig, ax = plt.subplots(figsize=(12, 5))

ax.plot([1, 2, 3, 4], [10, 20, 15, 25])
ax.set_title("Wide Figure")

plt.tight_layout()
plt.show()
````

Useful layout tools:

- `figsize=(width, height)`
- `plt.tight_layout()`
- `constrained_layout=True`

For more complex figures, `constrained_layout=True` can often produce better spacing.

````python
fig, axes = plt.subplots(
    2, 2,
    figsize=(10, 7),
    constrained_layout=True
)

for ax in axes.flat:
    ax.plot(np.arange(5), np.random.random(5))
    ax.grid(alpha=0.2)

plt.show()
````

# 22. Text and Annotations

````python
x = np.arange(1, 11)
y = np.array([10, 15, 13, 20, 25, 23, 30, 34, 32, 40])

fig, ax = plt.subplots()

ax.plot(x, y, marker="o")

max_index = y.argmax()

ax.annotate(
    "Highest value",
    xy=(x[max_index], y[max_index]),
    xytext=(6, 45),
    arrowprops=dict(arrowstyle="->")
)

ax.text(1, 35, "Important trend")

plt.show()
````

Use annotations to explain the story of a chart.

Good annotations answer questions such as:

- What is unusual here?
- When did something change?
- Which point matters?
- Where is the target?

# 23. Date and Time-Series Plots

````python
dates = pd.date_range("2026-01-01", periods=100, freq="D")
sales = 100 + np.cumsum(np.random.normal(0, 3, 100))

fig, ax = plt.subplots()

ax.plot(dates, sales)

ax.set_title("Daily Sales")
ax.set_xlabel("Date")
ax.set_ylabel("Sales")

fig.autofmt_xdate()

plt.show()
````

Matplotlib can work directly with Python datetime objects and Pandas timestamps.

For time series, pay attention to:

- date formatting
- tick frequency
- overlapping labels
- rolling averages
- seasonality

# 24. Formatting Axes

````python
from matplotlib.ticker import FuncFormatter, MultipleLocator

values = [1000, 2500, 5000, 7500, 10000]

fig, ax = plt.subplots()

ax.bar(range(len(values)), values)

ax.yaxis.set_major_formatter(
    FuncFormatter(lambda value, position: f"${value:,.0f}")
)

ax.yaxis.set_major_locator(MultipleLocator(2500))

ax.set_title("Revenue")
ax.set_xlabel("Category")

plt.show()
````

Formatting is especially important for:

- currency
- percentages
- thousands/millions
- dates
- scientific notation
- custom tick labels

# 25. Twin Axes

````python
months = np.arange(1, 7)
revenue = [100, 130, 160, 180, 220, 250]
customers = [20, 25, 27, 35, 40, 45]

fig, ax1 = plt.subplots()

ax1.plot(months, revenue, marker="o", label="Revenue")
ax1.set_xlabel("Month")
ax1.set_ylabel("Revenue")

ax2 = ax1.twinx()
ax2.plot(months, customers, marker="s", linestyle="--", label="Customers")
ax2.set_ylabel("Customers")

ax1.set_title("Revenue and Customers")

plt.show()
````

Twin axes can be useful when two variables have very different units.

However, they can also mislead readers. Use them sparingly and label both axes clearly.

# 26. Error Bars

````python
x = np.arange(5)
means = [10, 14, 12, 18, 20]
errors = [1, 2, 1.5, 2.5, 1.8]

fig, ax = plt.subplots()

ax.errorbar(
    x,
    means,
    yerr=errors,
    fmt="o-",
    capsize=5
)

ax.set_title("Measurements with Uncertainty")
ax.set_xlabel("Experiment")
ax.set_ylabel("Value")

plt.show()
````

Error bars can represent standard deviation, standard error, confidence intervals, or another uncertainty measure. Always explain what they represent.

# 27. Filled Areas

````python
x = np.linspace(0, 10, 200)
y = np.sin(x)

fig, ax = plt.subplots()

ax.plot(x, y)
ax.fill_between(x, y, 0, alpha=0.2)

ax.axhline(0, linestyle="--")

plt.show()
````

````python
x = np.arange(10)
actual = np.array([10, 12, 11, 15, 14, 18, 20, 19, 23, 25])
target = np.array([11, 11, 12, 13, 15, 16, 18, 20, 21, 22])

fig, ax = plt.subplots()

ax.plot(x, actual, label="Actual")
ax.plot(x, target, label="Target")
ax.fill_between(x, actual, target, alpha=0.2)

ax.legend()
plt.show()
````

# 28. Statistical Visualizations

````python
np.random.seed(42)

groups = {
    "Group A": np.random.normal(50, 8, 100),
    "Group B": np.random.normal(60, 10, 100),
    "Group C": np.random.normal(55, 6, 100)
}

fig, ax = plt.subplots()

ax.boxplot(groups.values())
ax.set_xticklabels(groups.keys())
ax.set_ylabel("Value")
ax.set_title("Group Distributions")

plt.show()
````

ultimately build on its concepts.

# 31. Visualization Best Practices

## 1. Start with the question

Do not start by asking "Which chart looks cool?"

Start with:

> What do I want the reader to understand?

## 2. Choose the chart based on the task

| Goal | Typical chart |
|---|---|
| Trend over time | Line |
| Compare categories | Bar |
| Distribution | Histogram / Box plot |
| Relationship | Scatter |
| Composition | Stacked bar / sometimes pie |
| Uncertainty | Error bars / shaded interval |

## 3. Label everything important

A reader should understand:

- what is being measured
- the units
- what the categories mean
- what the time period is

## 4. Avoid unnecessary decoration

3D effects, excessive grid lines, too many colors, and unnecessary borders can reduce clarity.

## 5. Use consistent scales

Changing scales between similar charts can make comparisons misleading.

## 6. Highlight the important information

If one point matters, annotate it.

## 7. Consider accessibility

Do not rely on color alone to communicate meaning.

## 8. Keep the audience in mind

An exploratory chart for yourself can be different from a chart in an executive report.

# Matplotlib Mental Model

Remember the hierarchy:

**Figure → Axes → Artists**

The most useful pattern is:

```python
fig, ax = plt.subplots()

ax.plot(...)
ax.set_title(...)
ax.set_xlabel(...)
ax.set_ylabel(...)
ax.legend(...)
ax.grid(...)

plt.show()
```

For multiple charts:

```python
fig, axes = plt.subplots(...)
```

For exporting:

```python
fig.savefig(...)
```

And the general visualization workflow is:

**QUESTION → DATA → PREPARE → CHOOSE CHART → BUILD → LABEL → INTERPRET → EXPORT**

The goal of visualization is not to create beautiful pictures.

The goal is to make the important information **easy to see and understand**.
