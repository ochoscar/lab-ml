# 2. Why NumPy?

Python lists are flexible, but numerical operations over large collections can be
slow and verbose.

NumPy provides:

- compact numerical arrays
- vectorized operations
- broadcasting
- fast mathematical functions
- multidimensional arrays
- linear algebra
- random number generation
- efficient memory usage

For example, with Python lists, multiplying every value requires iteration.
With NumPy, the operation can be expressed directly.

````python
values = [1, 2, 3, 4, 5]
array = np.array(values)

array * 10
````

The important idea is:

> **Describe the operation on the whole array instead of manually looping over every element.**

# 14. Aggregations by Axis

````python
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("Total:", matrix.sum())
print("Column sums:", matrix.sum(axis=0))
print("Row sums:", matrix.sum(axis=1))
````

````python
print("Column means:", matrix.mean(axis=0))
print("Row means:", matrix.mean(axis=1))
````

For a 2D array:

- `axis=0` aggregates **down rows**, producing one value per column.
- `axis=1` aggregates **across columns**, producing one value per row.

This is a critical concept in NumPy.

# 15. Broadcasting

Broadcasting allows NumPy to perform operations between arrays with compatible shapes.

Example:

```python
matrix + vector
```

The vector can be conceptually applied to every row.

````python
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

offset = np.array([1, 2, 3])

matrix + offset
````

````python
prices = np.array([
    [100, 200, 300],
    [150, 250, 350]
])

tax_rates = np.array([0.10, 0.20, 0.15])

prices * (1 + tax_rates)
````

Broadcasting generally works when dimensions are equal or one of them is 1.

A useful rule is to compare shapes from the **rightmost dimension backward**.

# 24. Random Simulations

NumPy is extremely useful for simulation.

For example, suppose a customer has a 20% probability of cancelling an order.
We can simulate many orders and estimate the observed cancellation rate.

````python
rng = np.random.default_rng(42)

n = 100_000
cancelled = rng.random(n) < 0.20

cancelled.mean()
````

With enough observations, the simulated rate should be close to the theoretical probability.

````python
rng = np.random.default_rng(123)

dice = rng.integers(1, 7, size=100_000)

print("Mean:", dice.mean())
print("Expected mean:", 3.5)
````

### Monte Carlo idea

Monte Carlo methods repeatedly simulate random outcomes to estimate probabilities,
expectations, or other quantities that may be difficult to calculate analytically.

# 26. Matrix Operations

````python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("Element-wise multiplication:")
print(A * B)

print("\nMatrix multiplication:")
print(A @ B)

print("\nUsing matmul:")
print(np.matmul(A, B))
````

This distinction is fundamental:

- `A * B` → element-by-element multiplication
- `A @ B` → matrix multiplication

Do not confuse the two.

````python
A = np.array([[3, 1], [1, 2]])
b = np.array([9, 8])

solution = np.linalg.solve(A, b)

print(solution)
print(A @ solution)
````

# 27. Practical Numerical Patterns

### Normalize values

A common min-max normalization is:

`(x - min) / (max - min)`

````python
values = np.array([10, 20, 30, 40, 50])

normalized = (values - values.min()) / (values.max() - values.min())
normalized
````

### Z-score standardization

````python
values = np.array([10, 20, 30, 40, 50])

z_scores = (values - values.mean()) / values.std()
z_scores
````

### Pairwise differences

````python
values = np.array([10, 13, 18, 20, 27])

np.diff(values)
````

### Cumulative operations

````python
values = np.array([10, 20, 5, 15])

print(np.cumsum(values))
print(np.cumprod(values))
````

# 28. Performance

One of NumPy's major advantages is vectorization.

Let's compare a Python loop with a NumPy operation.

````python
import time

n = 1_000_000
values = list(range(n))

start = time.perf_counter()
result_loop = [x * 2 for x in values]
loop_time = time.perf_counter() - start

array = np.arange(n)

start = time.perf_counter()
result_numpy = array * 2
numpy_time = time.perf_counter() - start

print(f"Python loop: {loop_time:.6f} seconds")
print(f"NumPy:       {numpy_time:.6f} seconds")
````

The exact speedup depends on hardware and the operation, but NumPy's vectorized
implementation can be dramatically faster for numerical workloads.

### General performance principles

- Prefer vectorized NumPy operations.
- Avoid Python loops over large numerical arrays.
- Use appropriate dtypes.
- Avoid unnecessary copies.
- Work with contiguous arrays when relevant.
- Profile before optimizing.

# 29. NumPy + Pandas
