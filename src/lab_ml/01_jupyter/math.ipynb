Simple Fourier serie w=w_0*x/L

```python
import numpy as np
import matplotlib.pyplot as plt
import math

L = 10.0
w0 = 1.0
N = 100
x_range = 5 * L

x = np.arange(-x_range, x_range, 0.01)
y_vect = []

for xi in x:
    yi = 0.0

    for n in range(1, N + 1):
        yi += (2 * w0 / (n * np.pi)) * (-1)**(n + 1) * np.sin(n * np.pi * xi / L)

    y_vect.append(yi)

y = np.array(y_vect)

plt.plot(x, y, label="Serie de Fourier")
plt.plot(x, w0 * x / L, "--", label="Función original")
plt.grid()
plt.legend()
plt.show()
```
